```python
import matplotlib.pyplot as plt
import general

# 1. Appel de la fonction avec le modele de relaxation impulsive
results = general.extend_halo(extension_mode='Impulsive')

# 2. Extraction des rayons et des densites depuis le dictionnaire renvoye
r = results['r']
rho_dm_initial = results['rho_dm_initial']
rho_dm_extended = results['rho_dm_extended']
rho_baryons = results['rho_baryons']

# 3. Creation du graphique en echelle log-log
plt.figure(figsize=(10, 10))

# Trace de la matiere noire initiale (par exemple profil NFW avec sa pointe centrale)
plt.loglog(r, rho_dm_initial, label='DM Initiale', color='black', linestyle='--')

# Trace de la matiere noire apres l'explosion (formation du coeur)
plt.loglog(r, rho_dm_extended, label='DM Etendue (Blowout)', color='red')

# Trace des baryons pour reference
plt.loglog(r, rho_baryons, label='Baryons post-explosion', color='blue', alpha=0.5)

# 4. Formatage du graphique
plt.xlabel(r'$r$ [Mpc/h]', fontsize=12)
plt.ylabel(r'Densite [$M_\odot / (\mathrm{Mpc/h})^3$]', fontsize=12)
plt.title("Effet de l'explosion centrale (Blowout) sur le profil de Matiere Noire")
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.show() 
```
