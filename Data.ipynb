```python
import matplotlib.pyplot as plt
import config
import interpolated 
import constantes
```

```python
dark_matter = interpolated.InterpolatedProfile(constantes.INTERPOLATED_DM)
baryons = interpolated.InterpolatedProfile(constantes.INTERPOLATED_BARYONS)

rho_baryons = baryons(config.R)
rho_dark_matter = dark_matter(config.R)
```

```python
plt.figure(figsize=(10, 10))
plt.loglog(config.R, rho_dark_matter, label='DM Initiale (NFW)', color='black')
plt.loglog(config.R, rho_baryons, label='Baryons', color='blue')
plt.xlabel(r'$r / r_{vir}$', fontsize=12)
plt.ylabel(r'Density [$M_\odot \mathrm{pc}^{-3}$]', fontsize=12)
plt.title(f'Interpolated profiles')
# plt.xlim(1e-6, 1)
# plt.ylim(1e-8, 1e4)
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.show()
```
